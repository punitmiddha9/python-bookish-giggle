```python
# File Operation- Read And Write Files
# File handling is a crucial part of any programming language. 
# Python provides built-in functions and methods to read from and write to files, both text and binary. 
# This lesson will cover the basics of file handling, including reading and writing text files and binary files.
```

```python
# Read a Whole File 

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
# Read a file Line by Line 

with open('example.txt', 'r') as file:
    for line in file:
        print(line)
```

```python
with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip()) # This strip removes the newline character
```

```python
# Writing a File (Overwriting)

with open('example.txt', 'w') as file:
    file.write('Hello World!!!\n')
    file.write('This is a new Line\n')
```

```python
# Writing a File (Without Overwriting)

with open('example.txt', 'a') as file:
    file.write('Appending the text using append mode.\n')
```

```python
# Writing List of lines to a file 

lines = ['First Line \n', 'Second Line \n', 'Third Line \n']

with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
# Binary Files 

# Writing to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
# Reading a binary file
with open('example.bin', 'rb') as file:
    content = file.read()
    print(content)
```

```python
### Read the content froma  source text fiile and write to a destination text file
# Copying a text file
with open('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
#Read a text file and count the number of lines, words, and characters.
# Counting lines, words, and characters in a text file
def count_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'example.txt'
lines, words, characters = count_text_file(file_path)
print(f'Lines: {lines}, Words: {words}, Characters: {characters}')
```

```python
# The w+ mode in Python is used to open a file for both reading and writing. 
# If the file does not exist, it will be created. 
# If the file exists, its content is truncated (i.e., the file is overwritten).
```

```python
### Writing and then reading a file

with open('write_and_read.txt','w+') as file:
    file.write("Hello world\n")
    file.write("This is a new line \n")

    ## Move the file cursor to the beginning
    file.seek(0)

    ## Read the content of the file
    content=file.read()
    print(content)
```
